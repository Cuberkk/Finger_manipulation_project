```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from utils.Contact_estimator import ContactEstimator
```

```python
USER = ['shuyuan','yizhou']
rotation = ['roll', 'pitch', 'yaw']
test_num = [1, 2, 3, 4, 5]
data_dir_dict = {}
save_dir_dict = {}
diameter = 50
CE = ContactEstimator(diameter=float(diameter))
for user in USER:
    data_dir_dict[user] = {}
    save_dir_dict[user] = {}
    for rot in rotation:
        data_dir_dict[user][rot] = []
        save_dir_dict[user][rot] = []
        for num in test_num:
            file_path = f'data/raw_data/{diameter}mm/{user}/{rot}/test_{num}/raw_data.csv'
            data_dir_dict[user][rot].append(file_path)
            file_save_path = f'data/raw_data/{diameter}mm/{user}/{rot}/test_{num}/transformed_data.csv'
            save_dir_dict[user][rot].append(file_save_path)
```

# Function Libs

```python
def load_data(file_path):
    data = pd.read_csv(file_path, header=None).to_numpy().reshape(-1,18)
    return data
```

# Main Analysis

```python
for user in USER:
    for rot in rotation:
        data_lis = data_dir_dict[user][rot]
        save_data_lis = save_dir_dict[user][rot]
        for i in data_lis:
            raw_data = load_data(i)
            transformed_data = np.zeros((raw_data.shape[0], raw_data.shape[1]))
            # print(raw_data.shape)
            for j in range(raw_data.shape[0]):
                contact_state = CE.estimate_contact_tri(raw_data[j, :])
                transformed_data[j, :] = contact_state
            save_path = save_data_lis[data_lis.index(i)]
            np.savetxt(save_path, transformed_data, delimiter=',')
```
